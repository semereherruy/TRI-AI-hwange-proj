### Uploading and Preparing Datasets

To resolve the `FileNotFoundError`, please upload the `.zip` archives and `.csv` files mentioned in your prompt. The following cell will guide you through the upload process and then automatically unzip the archives and move all files to the correct `/content/datasets` directory.

````python
from google.colab import files
import shutil

# Create the base directory if it doesn't exist
os.makedirs(BASE, exist_ok=True)

print("Please upload the following files (click 'Choose Files' below):")
print("- English_Amharic_dataset_for_toxicity_detection_ubuntu_hate_speech.csv")
print("- ubuntu_hate_speech_synthetic_dataset_en_am_500rows.csv")
print("- AfriHate-main.zip")
print("- HateXplain-master.zip")
print("- TOXIGEN-main.zip")

# Upload files
uploaded = files.upload()

# Process uploaded files
for filename in uploaded.keys():
    print(f'Processing uploaded file: "{filename}"')
    source_path = os.path.join("/content", filename)

    if filename.endswith(".zip"):
        # Unzip archives to /content first, then move the main extracted folder to BASE
        target_dir_name = filename.replace(".zip", "")
        extract_path = "/content"

        # Ensure destination in BASE doesn't conflict if a partial unzipping occurred
        destination_path_in_base = os.path.join(BASE, target_dir_name)
        if os.path.exists(destination_path_in_base):
            print(f"Removing existing directory {destination_path_in_base} before extracting.")
            shutil.rmtree(destination_path_in_base)

        shutil.unpack_archive(source_path, extract_path)

        # After unzipping, the content will be in /content/target_dir_name
        # Move this entire directory to /content/datasets/target_dir_name
        extracted_folder_path = os.path.join(extract_path, target_dir_name)
        if os.path.exists(extracted_folder_path):
            shutil.move(extracted_folder_path, destination_path_in_base)
            print(f"Moved '{extracted_folder_path}' to '{destination_path_in_base}'")
        else:
            print(f"Warning: Expected directory '{extracted_folder_path}' not found after unzipping '{filename}'. Please check the contents of the zip file.")

    elif filename.endswith(".csv"):
        # Move CSV files directly into BASE
        shutil.move(source_path, os.path.join(BASE, filename))
        print(f"Moved '{filename}' to '{BASE}'")

print("\nFile preparation complete. Please re-run the data loading cells below.")
````

#### HateXplain Dataset Inspection

````python
print("\n--- HateXplain Dataset ---")
print(f"Number of entries: {len(hatexplain_data)}")

# Convert to DataFrame for easier inspection of aggregated features
# This assumes 'post_tokens' can be joined into a string for text content
hatexplain_list = []
for post_id, data in hatexplain_data.items():
    entry = {
        'post_id': post_id,
        'text': ' '.join(data['post_tokens']),
        'rationales_count': len(data['rationales']),
        'annotator_labels': [ann['label'] for ann in data['annotators']],
        'annotator_targets': [ann['target'] for ann in data['annotators']]
    }
    hatexplain_list.append(entry)

hatexplain_df = pd.DataFrame(hatexplain_list)

print("\nDataFrame Head (first 5 rows):")
display(hatexplain_df.head())

print("\nShape:")
print(hatexplain_df.shape)

print("\nColumns:")
print(hatexplain_df.columns.tolist())

print("\nData Types:")
print(hatexplain_df.info())

print("\nMissing Values:")
print(hatexplain_df.isnull().sum())

# Analyze annotator labels (majority vote or all labels)
# For simplicity, let's get a flattened list of all labels to understand distribution
from collections import Counter
all_labels = [label for sublist in hatexplain_df['annotator_labels'] for label in sublist]
print("\nAggregated Label Distribution (from all annotators):")
print(Counter(all_labels))

# Check if targets are always 'None' or if they contain specific groups
all_targets = [target for sublist in hatexplain_df['annotator_targets'] for target_list in sublist for target in target_list]
print("\nAggregated Target Distribution (from all annotators):")
print(Counter(all_targets))

print("\nCheck for Duplicate Text:")
duplicate_text_hx = hatexplain_df[hatexplain_df.duplicated(subset=['text'], keep=False)]
print(f"Number of duplicate texts: {len(duplicate_text_hx)}")
if not duplicate_text_hx.empty:
    print(duplicate_text_hx.sort_values('text').head())
````
